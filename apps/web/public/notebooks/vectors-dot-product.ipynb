# ML Math Lab: Vectors and Dot Products
Represent three normalized house features, calculate their weighted contributions, and debug vector shape/order mistakes.

```python
import numpy as np

features = np.array([2, 3, 4], dtype=float)
weights = np.array([5, 2, 3], dtype=float)

contributions = features * weights
prediction = np.dot(features, weights)
print(contributions)  # [10.  6. 12.]
print(prediction)     # 28.0
```

## Debug the contract
What happens when a pipeline emits `[bedrooms, size, location]` while the weights expect `[size, bedrooms, location]`?

```python
wrong_order = np.array([3, 2, 4], dtype=float)
print(np.dot(wrong_order, weights))  # 31.0, silently wrong

# Fix the data contract, not just this row.
correct_order = wrong_order[[1, 0, 2]]
assert np.dot(correct_order, weights) == 28.0
```

```python
# Shape mismatch: every feature needs exactly one weight.
try:
    np.dot(np.array([2, 3, 4]), np.array([5, 2]))
except ValueError as error:
    print(error)
```
