# ML Math Lab: Matrices & Matrix Multiplication
Batch predictions, inspect shapes, and repair a feature-schema mismatch.

```python
import numpy as np

X = np.array([[2,3,4],[4,2,5],[3,4,2]], dtype=float)
w = np.array([5,2,3], dtype=float)
predictions = X @ w
print(predictions)
assert np.allclose(predictions, [28,39,29])
assert predictions.shape == (3,)
```

```python
# Exercise: add a fourth house.
X4 = np.vstack([X, [5,3,4]])
p4 = X4 @ w
assert np.allclose(p4, [28,39,29,43])
assert p4.shape == (4,)

# Change the size weight: every row with nonzero size changes.
w2 = w.copy(); w2[0] = 6
print(X4 @ w2)
```

```python
# Shape mismatch: the schema has 3 features but only 2 weights.
broken_w = np.array([5,2])
try:
    X @ broken_w
except ValueError as error:
    print(error)

# Semantic fix: restore the documented location weight.
fixed_w = np.array([5,2,3])
assert np.allclose(X @ fixed_w, [28,39,29])
```

```python
# Matrix-matrix: two output columns can represent two neurons.
W = np.array([[5,1],[2,-1],[3,2]])
Y = X @ W
print(Y)
assert Y.shape == (3,2)
```
