# Functions, Graphs & Derivatives
Connect local slope to machine-learning loss sensitivity.

```python
import numpy as np

def loss(w): return (2*w-6)**2
def analytical_derivative(w): return 4*(2*w-6)

w=2.0; h=1e-5
numerical=(loss(w+h)-loss(w-h))/(2*h)
print(analytical_derivative(w), numerical)
assert np.isclose(analytical_derivative(w), numerical, atol=1e-7)
```

```python
for w in [2.,3.,4.]:
    print(w, loss(w), analytical_derivative(w))
assert [analytical_derivative(w) for w in [2.,3.,4.]] == [-8.,0.,8.]
```

```python
# Debug: this misses the inner derivative of 2w-6.
def incorrect_derivative(w): return 2*(2*w-6)
assert not np.isclose(incorrect_derivative(2), numerical, atol=1e-7)
assert np.isclose(4*(2*w-6), analytical_derivative(w))
```
