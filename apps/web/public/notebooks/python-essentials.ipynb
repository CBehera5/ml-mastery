## P01 — Values, names, operators and units

````python
import importlib.metadata
import json
import random
import sys
assert sys.version_info[:3] == (3, 13, 16), 'Use the approved exact Python reference profile.'
versions = {'numpy': '2.5.4', 'pandas': '3.0.6', 'matplotlib': '3.11.2', 'scikit-learn': '1.9.1'}
assert {name: importlib.metadata.version(name) for name in versions} == versions
random.seed(42)
results = {'profile': versions, 'seed': 42}
````

## 1. Predict type and meaning before running
A postcode '003' is an identifier, not a room count. A measured area can be numeric; True/False expresses a condition. Changing representation changes available operations. Predict the types of 3, 2.5, '3' and True.

````python
values = [3, 2.5, '3', True]
types = [type(value).__name__ for value in values]
assert types == ['int', 'float', 'str', 'bool']
results['types'] = types
print(types)
````

## 2. Types, order and units
Area is in hundreds of square metres, weight in ₹ lakh per area unit, bias in ₹ lakh. Multiplication precedes addition. First predict area 2, then area 4. Why does doubling area not double the complete price? Explain why '2'*3 is not a numeric prediction.

````python
area = 2
weight = 100
bias = 50
prediction = area * weight + bias
changed_prediction = 4 * weight + bias
repeated_text = '2' * 3
assert prediction == 250 and changed_prediction == 450
assert repeated_text == '222' and type(repeated_text) is str
results['predictions_in_lakh_rupees'] = [prediction, changed_prediction]
results['repeated_text'] = repeated_text
print(prediction, changed_prediction, repeated_text)
````

## 3. Trace updates, then compare
Predict the name/value table after each line. Assignment stores a new value; == asks a Boolean question without changing the name's value.

````python
area = 2
area = area + 1
comparison = area == 2
assert area == 3 and comparison is False
results['final_area'] = area
results['comparison'] = comparison
print(area, comparison)
````

## 4. Repair grouping and conversion
The intended rule adds one area unit before multiplying by 100. Predict both equivalent repairs and contrast the wrong precedence. Then predict what converting 'unknown' does: an error is not a zero price.

````python
area = '2'
grouped = (float(area) + 1) * 100
distributed = float(area) * 100 + 100
wrong_precedence = float(area) + 1 * 100
assert grouped == distributed == 300.0 and wrong_precedence == 102.0
try:
    float('unknown')
except ValueError:
    conversion_outcome = 'ValueError'
else:
    raise AssertionError('Expected invalid conversion to fail')
results['equivalent_repairs'] = [grouped, distributed]
results['wrong_precedence'] = wrong_precedence
results['invalid_conversion'] = conversion_outcome
print(json.dumps(results, sort_keys=True))
````

### P03.O2 — Distinguish a legitimate zero from None in a data rule.
Predict this worked state/error before running. Zero is a legitimate observed value. Test is not None, rather than truthiness, so the zero-valued record is not silently lost.

Remediation: Compare records with 0, None and an ordinary positive value.

````python
records = [('A3', 0), ('B3', None), ('C3', 3)]
kept = [row_id for row_id, age in records if age is not None]
result = kept
normalized = json.loads(json.dumps(result))
assert normalized == json.loads("[\"A3\",\"C3\"]")
all_results.append({'objective': 'P03.O2', 'outcome': normalized})
print(json.dumps(normalized))
````

### P03.O3 — Trace mutually exclusive if/elif/else branches.
Predict this worked state/error before running. An if/elif chain executes the first satisfied branch; test the exact boundary before choosing a later branch.

Remediation: Annotate the first satisfied branch.

````python
age = 3
if age < 3:
    branch = 'low'
elif age <= 4:
    branch = 'middle'
else:
    branch = 'high'
result = branch
normalized = json.loads(json.dumps(result))
assert normalized == json.loads("\"middle\"")
all_results.append({'objective': 'P03.O3', 'outcome': normalized})
print(json.dumps(normalized))
````

### P03.O4 — Reject a condition that changes the intended population.
Predict this worked state/error before running. This truthiness filter accidentally drops valid zero A as well as missing C. A plausible row count does not prove the intended population was kept.

Remediation: Compare the retained row IDs, not just the count.

````python
records = [('A', 0), ('B', 3), ('C', None)]
kept = [row_id for row_id, age in records if age]
result = kept
normalized = json.loads(json.dumps(result))
assert normalized == json.loads("[\"B\"]")
all_results.append({'objective': 'P03.O4', 'outcome': normalized})
print(json.dumps(normalized))
````

## P04 — Loops and reliable aggregations
An aggregate carries forward all prior contributions and needs an honest denominator.

Mathematical connection: Finite sums, means and paired residuals.

Terminology: for, range, enumerate, zip, accumulator, count, loop invariant.

Common misconception: The final loop value is not the aggregate; zip stops at the shorter input.

Change the case: Start with three errors, then an empty batch and unequal paired lists.

Record outcomes are normalized JSON: Python True/False/None become true/false/null; tuple pairs retain order as arrays. The program's type names are retained explicitly when assessed.

### P04.O1 — Trace a bounded loop and half-open range.
Predict this worked state/error before running. range includes its start and excludes its stop; expand the visited indices before counting.

Remediation: Expand range into an explicit index list.

````python
visited = list(range(1, 5))
result = [visited, len(visited)]
normalized = json.loads(json.dumps(result))
assert normalized == json.loads("[[1,2,3,4],4]")
all_results.append({'objective': 'P04.O1', 'outcome': normalized})
print(json.dumps(normalized))
````

### P04.O2 — Compute an aggregate without dropping or resetting contributions.
Predict this worked state/error before running. Initialize once outside the loop and retain every contribution; the denominator counts all three observations, including zero error.

Remediation: Write one accumulator row per observation.

````python
errors = [-3, 0, 3]
total = 0
count = 0
for error in errors:
    total += error * error
    count += 1
result = [total, count, total / count]
normalized = json.loads(json.dumps(result))
assert normalized == json.loads("[18,3,6]")
all_results.append({'objective': 'P04.O2', 'outcome': normalized})
print(json.dumps(normalized))
````

### P04.O3 — Maintain row pairing with enumerate/zip and detect unequal lengths.
Predict this worked state/error before running. zip truncates at the shorter list. Check lengths and row identities before trusting paired model errors.

Remediation: Check input lengths and paired row IDs before iteration.

````python
features = [3, 4, 5]
targets = [100, 200]
pairs = list(zip(features, targets))
result = [pairs, len(features) == len(targets)]
normalized = json.loads(json.dumps(result))
assert normalized == json.loads("[[[3,100],[4,200]],false]")
all_results.append({'objective': 'P04.O3', 'outcome': normalized})
print(json.dumps(normalized))
````

### P04.O4 — Explain a zero-count division guard and bounded iteration.
Predict this worked state/error before running. An empty batch has no observed mean; guard zero count. Observed zero errors, unlike an empty batch, have a legitimate mean zero.

Remediation: Separate count validation from mean calculation.

````python
errors = [0, 0]
count = len(errors)
result = None if count == 0 else sum(errors) / count
normalized = json.loads(json.dumps(result))
assert normalized == json.loads("0")
all_results.append({'objective': 'P04.O4', 'outcome': normalized})
print(json.dumps(normalized))
````

## P05 — Functions turn repeated steps into contracts

### P05.O1 — Identify parameters, arguments and local variables in a supplied function.
Predict this worked state/error before running. area is the parameter, the supplied number is its argument, and price is local to the call frame.

Remediation: Trace a call frame separately from caller variables.

````python
def predict(area):
    price = area * 100 + 50
    return price
value = predict(3)
result = ['area', 3, 'price', value]
normalized = json.loads(json.dumps(result))
assert normalized == json.loads("[\"area\",3,\"price\",350]")
all_results.append({'objective': 'P05.O1', 'outcome': normalized})
print(json.dumps(normalized))
````

### P05.O2 — Return a usable result instead of only printing it.
Predict this worked state/error before running. Return the prediction so the next function receives a number. Printing without returning would pass None and break the arithmetic.

Remediation: Follow the returned value into the next calculation.

````python
def predict(area):
    return area * 100 + 50
def add_fee(price):
    return price + 10
result = add_fee(predict(3))
normalized = json.loads(json.dumps(result))
assert normalized == json.loads("360")
all_results.append({'objective': 'P05.O2', 'outcome': normalized})
print(json.dumps(normalized))
````

### P05.O3 — Compose small validation and prediction functions in order.
Predict this worked state/error before running. Validate and convert first, then predict using that numeric result. Composition carries the intermediate type and units forward.

Remediation: Draw the intermediate input/output types.

````python
def validate(text):
    return float(text)
def predict(area):
    return area * 100 + 50
result = predict(validate('3'))
normalized = json.loads(json.dumps(result))
assert normalized == json.loads("350")
all_results.append({'objective': 'P05.O3', 'outcome': normalized})
print(json.dumps(normalized))
````

### P05.O4 — Recognize mutable-default and unexpected-input mutation risks.
Predict this worked state/error before running. Create a fresh default on each call and copy supplied input before editing; neither a previous call nor caller-owned data should silently change.

Remediation: Compare successive calls and caller-owned data.

````python
def append_feature(value, items=None):
    output = [] if items is None else items.copy()
    output.append(value)
    return output
caller = [10]
first = append_feature(3)
second = append_feature(4)
third = append_feature(3, caller)
result = [first, second, third, caller]
normalized = json.loads(json.dumps(result))
assert normalized == json.loads("[[3],[4],[10,3],[10]]")
all_results.append({'objective': 'P05.O4', 'outcome': normalized})
print(json.dumps(normalized))
````

### P06.O1 — Translate a one-level loop into a list comprehension without changing order.

### P06.O2 — Preserve feature/target alignment when filtering records.

Remediation: Compare retained row IDs on both sides.

````python
records = [('A', 3, 100), ('B', None, 200), ('C', 4, 300)]
paired = [(row_id, area, target) for row_id, area, target in records if area is not None]
result = paired
normalized = json.loads(json.dumps(result))
assert normalized == json.loads("[[\"A\",3,100],[\"C\",4,300]]")
all_results.append({'objective': 'P06.O2', 'outcome': normalized})
print(json.dumps(normalized))
````

### P06.O3 — Predict a dictionary comprehension with duplicate keys.
Predict this worked state/error before running. A dictionary has one value per key; the later A insertion overwrites the earlier A, so duplicates can represent information loss.

Remediation: Trace each insertion into a keyed table.

````python
records = [('A', 3), ('A', 4), ('B', 9)]
result = {row_id: value for row_id, value in records}
normalized = json.loads(json.dumps(result))
assert normalized == json.loads("{\"A\":4,\"B\":9}")
all_results.append({'objective': 'P06.O3', 'outcome': normalized})
print(json.dumps(normalized))
````

### P06.O4 — Choose readable explicit code over an unsupported nested shortcut.
Predict this worked state/error before running. This readable loop maps each observation; an equivalent single-level comprehension produces the same list, not its sum.

Remediation: Separate map, filter and reduce stages.

````python
values = [3, 4]
output = []
for value in values:
    output.append(value * 2)
result = output
normalized = json.loads(json.dumps(result))
assert normalized == json.loads("[6,8]")
all_results.append({'objective': 'P06.O4', 'outcome': normalized})
print(json.dumps(normalized))
````

### P08.O1 — Validate a bundled record's schema, conversion and missingness before use.
Predict this worked state/error before running. Validate required fields, numeric conversion and nullability before modeling. Missing or invalid required area rejects the record; missing age must not become an observed zero.

Remediation: Apply the supplied field/type/nullability checklist.

````python
record = {'area': 'bad-3', 'age': ''}
age = None if record['age'] == '' else float(record['age'])
try:
    area = float(record['area'])
except (KeyError, ValueError):
    result = ['invalid required area', None, age]
else:
    result = ['valid', area, age]
normalized = json.loads(json.dumps(result))
assert normalized == json.loads("[\"invalid required area\",null,null]")
all_results.append({'objective': 'P08.O1', 'outcome': normalized})
print(json.dumps(normalized))
````

### P08.O2 — Explain context-managed reading and deterministic relative asset paths.
Predict this worked state/error before running. In this supplied program the context closes after parsing and validation. A declared relative bundled asset name is portable; an arbitrary machine-specific absolute path is not.

Remediation: Trace resource lifetime in a with block.

````python
import csv
import io
csv_text = 'area,age\n3,0\n'
asset_name = 'data/houses.csv'
with io.StringIO(csv_text) as handle:
    record = next(csv.DictReader(handle))
    valid = float(record['area']) > 0
result = [handle.closed, valid, asset_name, 'open-read/parse-validate-close']
normalized = json.loads(json.dumps(result))
assert normalized == json.loads("[true,true,\"data/houses.csv\",\"open-read/parse-validate-close\"]")
all_results.append({'objective': 'P08.O2', 'outcome': normalized})
print(json.dumps(normalized))
````

### P08.O4 — Recognize instance attributes and a simple method in a practical class (introductory).
Predict this worked state/error before running. Each instance owns its area attribute; changing first does not change second. A method uses its own instance through self; this is introductory recognition, not advanced OOP.

Remediation: Compare per-instance dictionaries; no inheritance required.

````python
class Record:
    def __init__(self, area):
        self.area = area
    def doubled(self):
        return self.area * 2
first = Record(3)
second = Record(4)
first.area = 5
result = [first.area, second.area, first.doubled()]
normalized = json.loads(json.dumps(result))
assert normalized == json.loads("[5,4,10]")
all_results.append({'objective': 'P08.O4', 'outcome': normalized})
print(json.dumps(normalized))
````
